# IoT Sensor Outlier Removal

**Task:** Clean sensor readings by removing rows with outliers in `temperature`, `humidity`, or `air_quality`.

**Method:** IQR — outlier if value < Q1 − 1.5×IQR  or  value > Q3 + 1.5×IQR

**Input:** (row_index, col_index) of cleaned dataframe  
**Output:** cell value at that location

```python
import pandas as pd

def remove_outliers(df):
    cols = ['temperature', 'humidity', 'air_quality']
    mask = pd.Series([True] * len(df), index=df.index)  # start: keep all rows

    for col in cols:
        Q1  = df[col].quantile(0.25)
        Q3  = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        mask = mask & df[col].between(lower, upper)  # keep only rows within bounds

    return df[mask].reset_index(drop=True)


# --- Load and clean ---
df = pd.read_csv('sensors.csv')
df_cleaned = remove_outliers(df)

# --- Answer test case ---
def answer(row_idx, col_idx):
    return df_cleaned.iloc[row_idx, col_idx]

print(df_cleaned.head())
```

```python
def clean_phone_numbers(contacts):
    cleaned = {}
    for name, number in contacts.items():
        digits = ''.join(ch for ch in number if ch.isdigit())
        cleaned[name] = digits if len(digits) == 10 else number
    return cleaned

# Test 1
t1 = {'Aisha': '(123) 456-7890', 'Bipin': '+123 654 3210', 'Nina': ' 000  '}
print(clean_phone_numbers(t1))
# Expected: {'Aisha': '1234567890', 'Bipin': '1236543210', 'Nina': ' 000  '}

# Test 2
t2 = {'Mansi': '123__ 9834 - 435', 'Tushar': '176 _ - 7576', 'Parul': '+1 (800) 555-1234'}
print(clean_phone_numbers(t2))
# Expected: {'Mansi': '1239834435', 'Tushar': '176 _ - 7576', 'Parul': '+1 (800) 555-1234'}
```
