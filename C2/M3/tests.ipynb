## Problem — Manufacturer's Claim (Product Life)

A manufacturer claims the average life of its product is **36 months**.  
An auditor selects a sample of **49 units**, calculates average life = **34.5 months**.  
Population standard deviation = **4 months**.  
Test at **α = 3%** using the critical value method.

```python
import math
from scipy import stats

# Given
mu0    = 36       # claimed population mean
n      = 49       # sample size
x_bar  = 34.5     # sample mean
sigma  = 4        # population SD
alpha  = 0.03     # significance level

# Hypotheses: H0: mu = 36, H1: mu != 36 (two-tailed — auditor tests the claim)

# Step 1 — Standard Error
SE = sigma / math.sqrt(n)
print(f"SE = {sigma} / √{n} = {SE:.4f}")

# Step 2 — Test statistic Z (from data)
Z = (x_bar - mu0) / SE
print(f"Z  = ({x_bar} - {mu0}) / {SE:.4f} = {Z:.4f}")

# Step 3 — Critical value Z* (from alpha, two-tailed)
# UCV cumulative = 1 - alpha/2
cum_prob = 1 - alpha / 2
z_star   = stats.norm.ppf(cum_prob)
print(f"\nTwo-tailed: alpha/2 = {alpha/2}, UCV cumulative = {cum_prob}")
print(f"Z* = {z_star:.4f}")

# Step 4 — Decision
print(f"\n|Z| = {abs(Z):.4f}  vs  Z* = {z_star:.4f}")
if abs(Z) > z_star:
    print("Decision: Reject H0 — insufficient evidence for manufacturer's claim")
else:
    print("Decision: Fail to reject H0 — claim holds")
```
