## Image Transformations

For various image analysis, image transformation techniques like rotation, mirroring, and various other transforms are important. One important application is in Data Augmentation concept used in Deep Learning. In this notebook, we will explore such fundamental image transformation techniques in Python.

What we will cover in this notebook:
- Image Rotation
- Mirroring
- Resizing
- Translation

### Setup

```python
import cv2
import numpy as np
import pandas as pda
import matplotlib.pyplot as plt
```

```python
image_path = 'data/car.png'
image = cv2.imread(image_path)
```

```python
# Utility function to display images
def show_image(img, fig_size = None):
    print(img.shape)
    img = img.astype('uint8')

    if fig_size is not None:
        plt.figure(figsize=fig_size)

    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.show()

show_image(image)
```

### Rotation
We will use `ndimage` from `scipy` library in Python to rotate our sample image data by a selected number of degrees.

```python
# For simplicity we will use scipy package
from scipy import ndimage

print("Rotate by 90 degress anti-clockwise")
show_image(ndimage.rotate(image,90))

print("Rotate by 90 degress clockwise")
show_image(ndimage.rotate(image,-90))
```

### Mirroring

Mirroring is like flipping the image from left to right or top to bottom and vice versa. We will use `flip()` from `numpy` array to use perform mirroring with images.

```python
print("Mirror horizontally")
show_image(np.flip(image,1))

print("Mirror Vertically")
show_image(np.flip(image,0))
```

### Resizing
We will use `resize()` from `open-cv` to resize our sample image. Resizing is commonly used in deep learning workflow, particularly when being inserted into the pipeline.

```python
show_image(cv2.resize(image,(50,50)))
```

### Translation
Image translation is defined as shifting an image or its contents by a fixed number of pixels horizontally (x-axis) or vertically (y-axis). It is often used in object tracking, basic image alignment, animation. We will use `warpAffine` from `open-cv` to perform this.

```python
M = np.float32([[1,0,80],[0,1,30]])
rows,cols = image.shape[0], image.shape[1]
show_image(cv2.warpAffine(image,M,(cols,rows)))
```

This brings us at the end of this notebook where we have performed fundamental image transformation methods using OpenCV, Python and other associated libraries.
