```python
import numpy as np

a = np.array([[10, 20, 30],
              [40, 99, 60],
              [70, 80, 90]])

flat = a.argmax()
pos  = np.unravel_index(flat, a.shape)

print(flat)   # 4
print(pos)    # (1, 1)
print(a[pos]) # 99
```

```python
from PIL import Image
import numpy as np

arr = np.random.randint(0, 256, (2,6,3), dtype=np.uint8)

arr[:, :, 0] = 0     # Red channel
arr[:, :, 1] = 255   # Green channel
arr[:, :, 2] = 2     # Blue channel
img = Image.fromarray(arr)
img                  # a green cube

# shape (3, 4, 3)  means:
#   3 rows
#   4 cols
#   3 values per pixel (R, G, B)

# row 0:  [R,G,B]  [R,G,B]  [R,G,B]  [R,G,B]
# row 1:  [R,G,B]  [R,G,B]  [R,G,B]  [R,G,B]
# row 2:  [R,G,B]  [R,G,B]  [R,G,B]  [R,G,B]
#          col 0    col 1    col 2    col 3
```

```python
import numpy as np
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


def show_image(img, fig_size = None):
    print(img.shape)
    img = img.astype('uint8')

    if fig_size is not None:
        plt.figure(figsize=fig_size)

    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.show()
```

```python
# create random noise image 300x400
arr = np.random.randint(0, 256, (300, 400, 3), dtype=np.uint8)
cv2.imshow('random noise', arr)
cv2.waitKey(0)
cv2.destroyAllWindows()


```

```python
arr[:, :, 0] = 0    # B
arr[:, :, 1] = 255  # G
arr[:, :, 2] = 0    # R

cv2.imshow('green', arr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
