# Dataset Preparation: Temporal Splits and Integrity Checks (MovieLens 1M)

Before we can train or evaluate any recommender models, we need to ensure that our dataset is **clean, consistent, and split properly over time.**
The goal of this notebook is to **prepare the MovieLens 1M dataset** so that it’s ready for real-world experiments — such as popularity baselines, collaborative filtering, matrix factorization, or learning-to-rank models.

## Connecting to the Real World

In real-world recommendation systems — think **Netflix recommending movies, Spotify suggesting songs, or Amazon showing related products** — data arrives over time, not all at once.
Each user’s actions form a **chronological journey**: what they watched, clicked, or rated earlier helps predict what they’ll do next.

If we mix past and future data or keep inconsistent user/movie records:
- Our model might “cheat” by seeing future interactions (data leakage).
- Metrics like Recall@K or NDCG would overestimate real-world performance.
- The business could make **poor recommendations** that don’t generalize to live users.

That’s why this step is critical — we’re simulating how recommendations are made **in production**, where the system only sees past behavior when predicting future actions.

## What We’re Achieving Here
- Build **time-aware splits** that mimic real production environments.
- Guarantee that all users, movies, and interactions are **valid and connected**.
- Convert raw ratings into **usable implicit feedback signals** (liked vs not liked).
- Prepare a **ready-to-train dataset** for modeling, evaluation, and serving experiments.

## Key Takeaways
- Learn how to structure a dataset for recommendation tasks that match business needs.
- Understand the importance of **temporal ordering and data integrity** in avoiding leakage.
- Produce clean, reproducible splits for accurate offline evaluation.
- End up with a dataset that can confidently drive downstream modeling and business experiments.

## Overall Agenda
1. **Configuration** – paths, parameters, and seed setup
1. **Data Loading & Schema Validation** – ensure columns and IDs are correct

1. **User–Item Integrity Analysis** – inspect activity patterns and long-tail behavior
1. **Split Strategy Definition** – decide how to partition user histories

1. **Split Summary & Next Steps** – review what we achieved and what comes next

## 1. Configuration

### 1.1 Paths
- Define data and output directories.
- Point to `ratings.csv`, `users.csv`, `movies.csv`.

```python
from pathlib import Path

DATA_DIR   = Path("data")
OUTPUT_DIR = Path("outputs")

RATINGS_PATH = DATA_DIR / "ratings.csv"
USERS_PATH   = DATA_DIR / "users.csv"
MOVIES_PATH  = DATA_DIR / "movies.csv"
```

### 1.2 Core Parameters
- Minimal knobs learners can understand and tweak.
- Balanced defaults for classroom scale.

```python
SEED = 42
K_VAL = 1                   # per-user last K → validation
K_TEST = 1                  # per-user last K → test
MIN_USER_INTERACTIONS = 3   # drop users with <3 ratings
MIN_ITEM_INTERACTIONS = 3   # drop movies with <3 ratings
RATING_THRESHOLD = 4.0      # rating ≥ 4 ⇒ liked (implicit)
TIMEZONE = "UTC"
```

### 1.3 Seed & Folders
- Ensure outputs directory exists.
- Set numpy/random seeds for repeatability.

```python
import random, numpy as np

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

def set_seed(s: int):
    random.seed(s)
    np.random.seed(s)

set_seed(SEED)
```

### 1.4 Input Presence Check
- Fail fast with a clear message if any file is missing.

```python
missing = [p for p in [RATINGS_PATH, USERS_PATH, MOVIES_PATH] if not p.exists()]
if missing:
    raise FileNotFoundError("Missing file(s):\n  " + "\n  ".join(str(p) for p in missing))

print("✅ Config ready\n")
print(f"DATA_DIR = {DATA_DIR}")
print(f"OUTPUT_DIR = {OUTPUT_DIR}")
print(f"K_VAL/K_TEST = {K_VAL}/{K_TEST}")
print(f"MIN_USER/ITEM = {MIN_USER_INTERACTIONS}/{MIN_ITEM_INTERACTIONS}")
print(f"RATING_THRESHOLD = {RATING_THRESHOLD}")
print(f"SEED = {SEED}")
print(f"TZ = {TIMEZONE}")
```

## 2. Data Loading & Schema Validation

### 2.1 Overview

In this step, we’ll load the **MovieLens 1M** dataset into Pandas and confirm that all tables have the correct structure.
We’ll make sure:
- Every file (ratings.csv, users.csv, movies.csv) is present and readable.
- Column names match the MovieLens schema.
- User and movie IDs are unique, with no duplicates.

This gives us a **clean foundation** before we start connecting tables or applying filters.

### 2.2 Why This Step Matters

In any real-world recommendation pipeline, data usually comes from multiple sources (user database, product catalog, event logs).
If the schemas don’t match or IDs are duplicated, the system can:
- fail to join data correctly,
- produce incorrect statistics, or
- silently leak inconsistent rows downstream.

So, this is where we build **trust in our data**.

### 2.3 Implementation Steps

#### 2.3.1 Load Each File
Load `ratings.csv`, `users.csv`, and `movies.csv` using Pandas.

```python
import pandas as pd

ratings = pd.read_csv(RATINGS_PATH)
users   = pd.read_csv(USERS_PATH)
movies  = pd.read_csv(MOVIES_PATH)

print("✅ Files loaded successfully!")
print(f"Ratings: {ratings.shape[0]:,} rows | Users: {users.shape[0]:,} | Movies: {movies.shape[0]:,}")
```

#### 2.3.2 Verify Column Names
Check that all expected columns are present exactly as in MovieLens 1M.

```python
ratings_cols = ["UserID", "MovieID", "Rating", "Timestamp"]
users_cols   = ["UserID", "Gender", "Age", "Occupation", "Zip-code"]
movies_cols  = ["MovieID", "Title", "Genres"]

def check_columns(df, expected, name):
    missing = [c for c in expected if c not in df.columns]
    if missing:
        raise ValueError(f"{name}.csv is missing columns: {missing}")

check_columns(ratings, ratings_cols, "ratings")
check_columns(users, users_cols, "users")
check_columns(movies, movies_cols, "movies")

print("✅ Column names verified for all files.")
```

#### 2.3.3 Check for Duplicate IDs
Confirm that each user and movie ID appears only once in its respective file.

```python
if users["UserID"].duplicated().any():
    raise ValueError("⚠️ Duplicate UserID found in users.csv")
if movies["MovieID"].duplicated().any():
    raise ValueError("⚠️ Duplicate MovieID found in movies.csv")

print("✅ No duplicate UserID or MovieID found.")
```

#### 2.3.4 Quick Preview
Display a few rows to visually confirm that the data looks as expected.

```python
print("\nRatings sample:")
display(ratings.head(3))

print("Users sample:")
display(users.head(3))

print("Movies sample:")
display(movies.head(3))
```

## 3. Referential Integrity Checks

### 3.1 Overview

We’ll verify that every rating in ratings.csv points to a **valid** user in `users.csv` and a **valid** movie in `movies.csv`.
<br>
Any _“orphan”_ ratings (missing user or movie) will be counted and removed.

### 3.2 Why This Step Matters
- Broken links cause **join failures** and **incorrect metrics** later.
- Cleaning now prevents silent bugs during splitting, baselines, and model training.
- Guarantees that each interaction is meaningful: `(UserID, MovieID)` exists in metadata.

### 3.3 Implementation Steps

#### 3.3.1 Build Membership Masks
Create boolean masks showing which ratings have valid users and valid movies.

```python
user_ok = ratings["UserID"].isin(users["UserID"])
movie_ok = ratings["MovieID"].isin(movies["MovieID"])
ok_mask = user_ok & movie_ok
```

#### 3.3.2 Count Orphans (Diagnostics)
Report how many ratings reference missing users or movies.

```python
n_before = len(ratings)
missing_users = (~user_ok).sum()
missing_movies = (~movie_ok).sum()

print("🔎 Referential diagnostics")
print(f"- Ratings total (before): {n_before:,}")
print(f"- Ratings with missing UserID : {missing_users:,}")
print(f"- Ratings with missing MovieID: {missing_movies:,}")
```

#### 3.3.3 Filter Out Invalid Ratings
Keep only rows that pass both checks; summarize the result.

```python
if missing_users or missing_movies:
    ratings = ratings.loc[ok_mask].copy()
    removed = n_before - len(ratings)
    print(f"⚠️ Removed {removed:,} invalid rating(s).")

n_after = len(ratings)
print(f"✅ Ratings total (after) : {n_after:,}")
print(f"✅ Unique users (ratings): {ratings['UserID'].nunique():,}")
print(f"✅ Unique movies (ratings): {ratings['MovieID'].nunique():,}")
```

## 4. Timestamp Validation & Range Profiling

### 4.1 Overview

Each rating in MovieLens includes a Timestamp showing when the user rated a movie.
In this step, we’ll make sure all timestamps:
- are valid Unix seconds,
- fall within a realistic date range, and
- make chronological sense for later temporal splits.

We’ll also convert timestamps into a readable date–time format for quick inspection.

### 4.2 Why This Step Matters
- Temporal order defines how we’ll split data into train, validation, and test later.
- If timestamps are corrupted or out of range, our model could “see the future.”
- Clean timestamps ensure a leak-free pipeline that mimics real-world recommendations.

### 4.3 Implementation Steps

#### 4.3.1 Convert Timestamps to Datetime
Convert the Unix Timestamp column to human-readable UTC datetimes.

```python
import pandas as pd

# Convert to datetime (UTC) for readability
ratings["Datetime"] = pd.to_datetime(ratings["Timestamp"], unit="s", utc=True)

# Quick sanity check
print("🕒 Example timestamps:")
display(ratings[["UserID", "MovieID", "Timestamp", "Datetime"]].head(3))
```

#### 4.3.2 Profile the Overall Time Range
Check the minimum and maximum timestamps and verify they look realistic.

```python
t_min, t_max = ratings["Datetime"].min(), ratings["Datetime"].max()
print(f"Dataset time range → {t_min.date()} to {t_max.date()}")
```

#### 4.3.3 Detect and Handle Outliers
MovieLens 1M should roughly lie between 1995 and 2005.
We’ll flag and drop anything outside this window.

```python
lower_bound = pd.Timestamp("1995-01-01", tz="UTC")
upper_bound = pd.Timestamp("2005-12-31", tz="UTC")

outliers = (~ratings["Datetime"].between(lower_bound, upper_bound)).sum()
print(f"Out-of-range timestamps: {outliers:,}")

if outliers > 0:
    ratings = ratings.loc[ratings["Datetime"].between(lower_bound, upper_bound)].copy()
    print(f"⚠️ Removed {outliers:,} rows with invalid timestamps.")
```

#### 4.3.4 Summarize Timestamp Profile
Summarize users, movies, and total ratings after timestamp validation.

```python
print("\n✅ Timestamp validation complete.")
print(f"Users : {ratings['UserID'].nunique():,}")
print(f"Movies: {ratings['MovieID'].nunique():,}")
print(f"Ratings: {len(ratings):,}")
print(f"Final time span: {ratings['Datetime'].min().date()} → {ratings['Datetime'].max().date()}")
```

## 5. Interaction Filtering & Label Preparation

### 5.1 Overview
This step cleans up the interaction data and adds a **binary feedback signal** `(label)` that tells us whether a user “liked” a movie.
We’ll also remove very **sparse users and movies** — those with too few ratings — because they can’t provide enough context for reliable evaluation.

### 5.2 Why This Step Matters
- In real recommendation systems, most feedback is **implicit** (clicks, views, purchases).
- Converting explicit ratings into implicit “likes” lets us model user preference behavior realistically.
- Removing sparse users/movies avoids skewing the dataset with noise or unreliable statistics.

### 5.3 Implementation Steps

#### 5.3.1 Create a Binary Label
Mark a rating as **liked (1)** if it meets or exceeds our threshold (e.g., ≥4).

```python
# Ensure Rating is numeric (in case it's read as string)
ratings["Rating"] = pd.to_numeric(ratings["Rating"], errors="coerce")

# Drop any rows where Rating could not be converted
before = len(ratings)
ratings = ratings.dropna(subset=["Rating"]).copy()
after = len(ratings)
if before != after:
    print(f"⚠️ Dropped {before - after} rows with invalid or missing ratings.")

# Convert explicit ratings to implicit binary labels
ratings["label"] = (ratings["Rating"].to_numpy() >= RATING_THRESHOLD)

# Quick summary
pos = int(ratings["label"].sum())
neg = len(ratings) - pos
print(f"👍 Positive (liked) ratings : {pos:,}")
print(f"👎 Negative (not liked)     : {neg:,}")
print(f"✅ Label column created successfully.")
```

#### 5.3.2 Remove Sparse Users and Movies
Filter out users and movies with too few interactions to ensure each has enough data for temporal splitting.

```python
before_rows = len(ratings)
before_users = ratings["UserID"].nunique()
before_movies = ratings["MovieID"].nunique()

# Count how many interactions each user/movie has
user_counts = ratings["UserID"].value_counts()
movie_counts = ratings["MovieID"].value_counts()

# Keep only those with enough interactions
keep_users = user_counts[user_counts >= MIN_USER_INTERACTIONS].index
keep_movies = movie_counts[movie_counts >= MIN_ITEM_INTERACTIONS].index

ratings = ratings[
    ratings["UserID"].isin(keep_users) & ratings["MovieID"].isin(keep_movies)
].copy()

after_rows = len(ratings)
print(f"Rows before/after filtering: {before_rows:,} → {after_rows:,}")
print(f"Users before/after: {before_users:,} → {ratings['UserID'].nunique():,}")
print(f"Movies before/after: {before_movies:,} → {ratings['MovieID'].nunique():,}")
```

#### 5.3.3 Sanity Check
Ensure required columns exist and the filtered dataset still looks healthy.

```python
expected_cols = {"UserID", "MovieID", "Rating", "Timestamp", "label"}
missing_cols = expected_cols - set(ratings.columns)
assert not missing_cols, f"Missing columns: {missing_cols}"

print("✅ Labeling and filtering complete. Dataset ready for integrity analysis next.")
```

## 6. User–Item Integrity Analysis

### 6.1 Overview

Before we create train/validation/test splits, we need to understand how our users and movies behave.
This step checks for:
- Users with **too few or too many interactions**.
- Movies that dominate or are rarely rated.
- The **time span** of each user’s activity (to confirm data continuity).

These checks help ensure our splits will be **balanced and realistic**.

### 6.2 Why This Step Matters
- Some users rate hundreds of movies, while others rate only a few.
- Some movies are extremely popular — these can bias the recommender if unchecked.
- Knowing the **activity span** per user helps us decide temporal split parameters (like how many ratings we can safely reserve for validation/test).

This is a quick **sanity and exploration** step before we define the splitting strategy.

### 6.3 Implementation Steps

#### 6.3.1 User-Level Interaction Summary
Compute per-user counts and activity spans.

```python
# Count number of interactions per user
user_activity = (
    ratings.groupby("UserID")
    .agg(
        n_ratings=("MovieID", "count"),
        first_time=("Datetime", "min"),
        last_time=("Datetime", "max")
    )
    .reset_index()
)

# Add user active duration (in days)
user_activity["active_days"] = (user_activity["last_time"] - user_activity["first_time"]).dt.days

print("User-level summary:")
display(user_activity.describe().T)
```

#### 6.3.2 Movie-Level Interaction Summary
Check how popular each movie is and whether there’s a strong “long-tail” effect.

```python
movie_activity = (
    ratings.groupby("MovieID")
    .agg(
        n_ratings=("UserID", "count"),
        first_time=("Datetime", "min"),
        last_time=("Datetime", "max")
    )
    .reset_index()
)

print("Movie-level summary:")
display(movie_activity.describe().T)
```

#### 6.3.3 Quick Popularity Distribution
Visualize how many movies fall into different popularity ranges.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
movie_activity["n_ratings"].hist(bins=50)
plt.xlabel("Number of ratings per movie")
plt.ylabel("Count of movies")
plt.title("Distribution of Movie Popularity (Long-Tail View)")
plt.show()
```

#### 6.3.4 Identify Extremes
Identify highly active users and extremely popular movies to understand skew.

```python
# Top 5 most active users
top_users = user_activity.nlargest(5, "n_ratings")[["UserID", "n_ratings", "active_days"]]
print("Top 5 active users:")
display(top_users)

# Top 5 most rated movies
top_movies = movie_activity.nlargest(5, "n_ratings")[["MovieID", "n_ratings"]]
print("Top 5 popular movies:")
display(top_movies)
```

## 7. Split Strategy Definition

### 7.1 Overview

Now that our dataset is clean and validated, we need to decide **how to divide each user’s movie-watching history** into splits that reflect real-world usage.
<br>
We’ll use a **per-user temporal split**, where:
- The earliest interactions go into **train**,
- The next few go into **validation**,
- The most recent ones become **test**.

This way, the model only trains on past data to predict future actions — just like in production systems.

### 7.2 Why This Step Matters

In recommender systems, **data leakage** (using future interactions in training) can make evaluation metrics meaningless.
Defining a proper strategy ensures:
- Train data contains only historical interactions.
- Validation/test sets represent realistic “next-item” or “future preference” scenarios.
- Consistency across experiments and learners.

### 7.3 Implementation Steps

#### 7.3.1 Define Parameters for Temporal Splitting
We’ll use `K_VAL` and `K_TEST` (defined in configuration) to decide how many latest interactions per user are reserved for validation and test.

```python
# Example of parameters set earlier
print(f"K_VAL = {K_VAL}, K_TEST = {K_TEST}")
print("Each user's last K_VAL interactions → validation, last K_TEST → test.")
```

#### 7.3.2 Sort Ratings by Time
Ensure every user’s interactions are sorted chronologically before splitting.

```python
ratings = ratings.sort_values(["UserID", "Datetime"]).reset_index(drop=True)
```

#### 7.3.3 Group and Assign Split Labels
For each user, assign “train”, “val”, or “test” tags based on interaction order.

##### 7.3.3.1 Ensure Chronological Order per User

```python
# Sort by user and time so "last K" truly means most recent interactions
ratings = ratings.sort_values(["UserID", "Datetime"]).reset_index(drop=True)
```

##### 7.3.3.2 Compute Per-User Positions and Group Sizes

```python
# 0-based rank of each interaction within a user's timeline
rank_in_user = ratings.groupby("UserID").cumcount()

# Total number of interactions for each user's row
n_in_user = ratings.groupby("UserID")["MovieID"].transform("size")
```

##### 7.3.3.3 Identify Users With Insufficient History

```python
# Identify users who don't have enough interactions
# Users with ≤ (K_VAL + K_TEST) interactions will stay entirely in the 'train' split.

g_sizes = ratings.groupby("UserID").size()                     # user → total interaction count
sufficient_users = g_sizes[g_sizes > (K_VAL + K_TEST)].index    # users eligible for val/test
insufficient_users = g_sizes[g_sizes <= (K_VAL + K_TEST)].index # users kept only in train

print(f"Total users: {len(g_sizes):,}")
print(f"Sufficient users (eligible for val/test): {len(sufficient_users):,}")
print(f"Insufficient users (train-only): {len(insufficient_users):,}")
```

##### 7.3.3.4 Build Masks for Test and Validation

```python
# Create logical masks for which interactions become part of the val/test sets.
# Works seamlessly with the groupby().tail() approach from the previous step.

# Subset: only users with sufficient history
sufficient_mask = ratings["UserID"].isin(sufficient_users)
subset = ratings.loc[sufficient_mask]

# Test indices → last K_TEST interactions per user
test_idx = subset.groupby("UserID", group_keys=False).tail(K_TEST).index

# Validation indices → the K_VAL interactions just before the test ones
val_idx = (
    subset.groupby("UserID", group_keys=False)
          .tail(K_VAL + K_TEST).index
          .difference(test_idx)
)

print(f"Test interactions: {len(test_idx):,}")
print(f"Validation interactions: {len(val_idx):,}")
```

##### 7.3.3.5 Assign Split Labels

```python
# Default all interactions to 'train'
ratings["split"] = "train"

# Assign validation and test using the precomputed indices
ratings.loc[val_idx,  "split"] = "val"
ratings.loc[test_idx, "split"] = "test"

# Quick check
print(ratings["split"].value_counts())
```

##### 7.3.3.6 Quick Distribution Check

```python
print("Split distribution:")
display(ratings["split"].value_counts())
```

##### 7.3.3.7 Sanity: Chronology Within Each User


```python
# Train must be earlier than Val (where both exist)
ok_train_val = (
    ratings.loc[ratings["split"]=="train"]
           .groupby("UserID")["Datetime"].max()
    <=
    ratings.loc[ratings["split"]=="val"]
           .groupby("UserID")["Datetime"].min()
)

# Val must be earlier than Test (where both exist)
ok_val_test = (
    ratings.loc[ratings["split"]=="val"]
           .groupby("UserID")["Datetime"].max()
    <=
    ratings.loc[ratings["split"]=="test"]
           .groupby("UserID")["Datetime"].min()
)

# Reindex with True for users missing either side, so they don't fail the check
ok_train_val = ok_train_val.reindex(ratings["UserID"].unique(), fill_value=True).all()
ok_val_test  = ok_val_test.reindex(ratings["UserID"].unique(), fill_value=True).all()

assert ok_train_val, "Train must be earlier than Val for each user (where both exist)."
assert ok_val_test,  "Val must be earlier than Test for each user (where both exist)."

print("✅ Per-user temporal order verified (train < val < test).")
```

#### 7.3.4 Sanity Check of Split Distribution
Let’s check how many interactions fall into each split.

```python
split_counts = ratings["split"].value_counts()
print("Split distribution:")
display(split_counts)
```

## 8. Temporal Split Construction

### 8.1 Overview

Now that each interaction is tagged as `train`/`val`/`test`, we’ll materialize the three split DataFrames we’ll use downstream.

### 8.2 Why This Step Matters
- Keeps each split isolated and easy to pass into later notebooks.
- Allows quick stats and sanity checks per split.
- Ensures we only keep the columns we actually need.

### 8.3 Implementation Steps

#### 8.3.1 Extract Split DataFrames

```python
train_df = ratings.loc[ratings["split"] == "train"].copy()
val_df   = ratings.loc[ratings["split"] == "val"].copy()
test_df  = ratings.loc[ratings["split"] == "test"].copy()

print(f"train: {len(train_df):,} | val: {len(val_df):,} | test: {len(test_df):,}")
```

#### 8.3.2 Ensure No Overlap (Pairs should be unique across splits)

```python
def pairs(df):
    return set(zip(df["UserID"], df["MovieID"], df["Timestamp"]))

p_train, p_val, p_test = pairs(train_df), pairs(val_df), pairs(test_df)

assert p_train.isdisjoint(p_val),  "Overlap found between train and val!"
assert p_train.isdisjoint(p_test), "Overlap found between train and test!"
assert p_val.isdisjoint(p_test),   "Overlap found between val and test!"

print("✅ No overlap of (UserID, MovieID, Timestamp) across splits.")
```

#### 8.3.3 Chronology Sanity (per user, where applicable)

```python
import pandas as pd

def max_dt(df): return df["Datetime"].max() if len(df) else pd.NaT
def min_dt(df): return df["Datetime"].min() if len(df) else pd.NaT

by_u = ratings.groupby("UserID")

ok_train_val = []
ok_val_test  = []

for uid, g in by_u:
    g = g.sort_values("Datetime")
    t_max = max_dt(g[g["split"] == "train"])
    v_min = min_dt(g[g["split"] == "val"])
    v_max = max_dt(g[g["split"] == "val"])
    s_min = min_dt(g[g["split"] == "test"])
    ok_train_val.append(pd.isna(t_max) or pd.isna(v_min) or (t_max <= v_min))
    ok_val_test.append(pd.isna(v_max) or pd.isna(s_min) or (v_max <= s_min))

assert all(ok_train_val), "Expected train to come before val for users that have both."
assert all(ok_val_test),  "Expected val to come before test for users that have both."

print("✅ Per-user chronology holds (train < val < test).")
```

#### 8.3.4 Keep Minimal Columns (clean, eval-ready)

```python
cols = ["UserID", "MovieID", "Rating", "Timestamp", "Datetime", "label"]

train_df = train_df[cols].reset_index(drop=True)
val_df   = val_df[cols].reset_index(drop=True)
test_df  = test_df[cols].reset_index(drop=True)

print("\nHead of Train DF")
print(train_df.head(2))

print("\nHead of Validation DF")
print(val_df.head(2))

print("\nHead of Test DF")
print(test_df.head(2))
```

#### 8.3.5 Quick Split Stats

**Positive interaction (POS)**: any user–item event labeled as liked, clicked, or high rating (label = 1).

```python
def split_stats(df, name):
    return {
        "name": name,
        "rows": len(df),
        "users": df["UserID"].nunique(),
        "movies": df["MovieID"].nunique(),
        "start": df["Datetime"].min(),
        "end": df["Datetime"].max(),
        "pos_rate": float(df["label"].mean()) if "label" in df else None,
    }

stats = [split_stats(train_df, "train"),
         split_stats(val_df,   "val"),
         split_stats(test_df,  "test")]

for s in stats:
    print(f"{s['name']:>5} | rows={s['rows']:,} | users={s['users']:,} | movies={s['movies']:,} "
          f"| {str(s['start'])[:10]} → {str(s['end'])[:10]} | pos_rate={s['pos_rate']:.3f}")
```

## 9. Warm/Cold User–Item Handling

### 9.1 Overview

We’ll ensure that evaluation focuses on **warm users** (appear in train) and decide how to treat **cold items** (first appear in val/test).
You can either **drop** cold cases for a clean warm-warm eval, or **tag** them to report separately.

### 9.2 Why This Step Matters
- Many offline setups assume the model has seen the **user’s history** during training.
- Items that first appear only in val/test create **cold-start** scenarios that some models can’t handle.
- Being explicit keeps metrics honest and comparable.

### 9.3 Implementation Steps

#### 9.3.1 Identify Warm Users and Items (from train)

```python
users_train = set(train_df["UserID"].unique())
items_train = set(train_df["MovieID"].unique())

print(f"Warm users (train): {len(users_train):,}")
print(f"Warm items (train): {len(items_train):,}")
```

#### 9.3.2 Tag Cold Users/Items in Val/Test

```python
# Add flags for analysis; we won't drop yet
for df in (val_df, test_df):
    df["cold_user"] = ~df["UserID"].isin(users_train)
    df["cold_item"] = ~df["MovieID"].isin(items_train)

print("Val cold_user %:", round(100 * val_df["cold_user"].mean(), 2))
print("Val cold_item %:", round(100 * val_df["cold_item"].mean(), 2))
print("Test cold_user %:", round(100 * test_df["cold_user"].mean(), 2))
print("Test cold_item %:", round(100 * test_df["cold_item"].mean(), 2))
```

#### 9.3.3 Enforce Warm-User Evaluation (recommended)

```python
# Keep only users present in train for val/test (common offline practice)
val_warm  = val_df.loc[~val_df["cold_user"]].copy()
test_warm = test_df.loc[~test_df["cold_user"]].copy()

print(f"val:  {len(val_df):,} → {len(val_warm):,} (warm users only)")
print(f"test: {len(test_df):,} → {len(test_warm):,} (warm users only)")
```

#### 9.3.4 Drop cold items for a strict warm-warm eval

```python
val_final  = val_warm.loc[~val_warm["cold_item"]].copy()
test_final = test_warm.loc[~test_warm["cold_item"]].copy()

print(f"val (warm-warm):  {len(val_warm):,} → {len(val_final):,}")
print(f"test (warm-warm): {len(test_warm):,} → {len(test_final):,}")
```

#### 9.3.5 Snapshot Split Sizes After Dropping Cold Items

```python
def brief(df, name):
    return (name, len(df), df["UserID"].nunique(), df["MovieID"].nunique())

for name, df in [
    ("train", train_df),
    ("val_final", val_final),
    ("test_final", test_final),
]:
    nm, n, u, i = brief(df, name)
    print(f"{nm:10s} | rows={n:,} | users={u:,} | items={i:,}")
```

## 10. Post-Split Validation

### 10.1 Overview

We’ll sanity-check that our `train_df`, `val_final`, and `test_final` are:
- **Overlap-free** (no duplicate interactions across splits),
- **Chronologically ordered per user** (train < val < test),
- **Consistent with warm/cold policy**, and
- **Reasonable in distribution** (size, users/items, positive rate).

### 10.2 Why This Step Matters
- Prevents subtle **leakage** that can inflate metrics.
- Ensures downstream baselines/MF/LTR evaluate on **realistic future data**.
- Catches data pipeline bugs **before** modeling.

### 10.3 Implementation Steps

#### 10.3.1 No Overlap Across Splits

```python
def triplets(df):
    return set(zip(df["UserID"], df["MovieID"], df["Timestamp"]))

p_tr, p_va, p_te = triplets(train_df), triplets(val_final), triplets(test_final)

assert p_tr.isdisjoint(p_va),  "Overlap between train and val!"
assert p_tr.isdisjoint(p_te),  "Overlap between train and test!"
assert p_va.isdisjoint(p_te),  "Overlap between val and test!"
print("✅ No (UserID, MovieID, Timestamp) overlap across splits.")
```

#### 10.3.2 Chronology per User (train < val < test)

```python
import pandas as pd

# Combine all splits to check user timelines
combined = pd.concat([
    train_df.assign(split="train"),
    val_final.assign(split="val"),
    test_final.assign(split="test")
])
by_u = combined.groupby("UserID")

ok_train_val, ok_val_test = [], []

for uid, g in by_u:
    g = g.sort_values("Datetime")

    # Extract boundary timestamps for each split
    t_max = g.loc[g["split"]=="train", "Datetime"].max()
    v_min = g.loc[g["split"]=="val", "Datetime"].min()
    v_max = g.loc[g["split"]=="val", "Datetime"].max()
    s_min = g.loc[g["split"]=="test", "Datetime"].min()

    # Validate temporal ordering
    ok_train_val.append(pd.isna(t_max) or pd.isna(v_min) or (t_max <= v_min))
    ok_val_test.append(pd.isna(v_max) or pd.isna(s_min) or (v_max <= s_min))

# Assert all users follow train < val < test order
assert all(ok_train_val), "Train must be earlier than Val for some users."
assert all(ok_val_test),  "Val must be earlier than Test for some users."

print("✅ Per-user chronology holds (train < val < test).")
```

#### 10.3.3 Warm Policy Check (users/items seen in train)

```python
users_tr = set(train_df["UserID"])
items_tr = set(train_df["MovieID"])

# Users: we enforced warm users in Step 9; verify
assert (~val_final["UserID"].isin(users_tr)).sum() == 0, "Found cold users in val_final!"
assert (~test_final["UserID"].isin(users_tr)).sum() == 0, "Found cold users in test_final!"
print("✅ No cold users in val/test (warm-user eval).")

# Items: policy may vary; here we just report (you decided in Step 9)
print("val cold-item %:", round(100 * (~val_final["MovieID"].isin(items_tr)).mean(), 2))
print("test cold-item %:", round(100 * (~test_final["MovieID"].isin(items_tr)).mean(), 2))
```

#### 10.3.4 Duplicate Rows Inside Each Split

```python
def has_dupe_rows(df):
    return df.duplicated(subset=["UserID", "MovieID", "Timestamp"]).any()

assert not has_dupe_rows(train_df), "Duplicates inside train!"
assert not has_dupe_rows(val_final), "Duplicates inside val_final!"
assert not has_dupe_rows(test_final), "Duplicates inside test_final!"
print("✅ No per-split duplicate interactions.")
```

#### 10.3.5 Basic Distribution Drift (sanity)

```python
def split_summary(df, name):
    return {
        "name": name,
        "rows": len(df),
        "users": df["UserID"].nunique(),
        "items": df["MovieID"].nunique(),
        "start": df["Datetime"].min(),
        "end": df["Datetime"].max(),
        "pos_rate": float(df["label"].mean()) if "label" in df else None
    }

sum_tr = split_summary(train_df, "train")
sum_va = split_summary(val_final, "val")
sum_te = split_summary(test_final, "test")

for s in (sum_tr, sum_va, sum_te):
    print(f"{s['name']:>5} | rows={s['rows']:,} | users={s['users']:,} | items={s['items']:,} "
          f"| {str(s['start'])[:10]} → {str(s['end'])[:10]} | pos_rate={s['pos_rate']:.3f}")
```

#### 10.3.6 Referential Integrity per Split (belt-and-suspenders)

```python
users_all  = set(users["UserID"])
movies_all = set(movies["MovieID"])

for name, df in [("train", train_df), ("val", val_final), ("test", test_final)]:
    miss_u = (~df["UserID"].isin(users_all)).sum()
    miss_i = (~df["MovieID"].isin(movies_all)).sum()
    assert miss_u == 0 and miss_i == 0, f"{name}: found orphan references!"
print("✅ Referential integrity holds within each split.")
```
