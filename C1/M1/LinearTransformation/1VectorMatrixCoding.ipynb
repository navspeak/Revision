````python
import numpy as np
````

## 1. Vectors

````python
# Creating vectors
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("a =", a)
print("b =", b)
print("shape:", a.shape)
````

````python
# Vector operations
print("Addition:            ", a + b)
print("Subtraction:         ", a - b)
print("Scalar multiply (3a):", 3 * a)
````

````python
# Dot product
dot = np.dot(a, b)
print("Dot product a·b:", dot)

# Same using @
print("Using @:        ", a @ b)
````

````python
# Norm (magnitude)
norm_a = np.linalg.norm(a)
print("||a|| =", norm_a)

# Angle between a and b
cos_theta = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
angle = np.degrees(np.arccos(cos_theta))
print(f"Angle between a and b: {angle:.2f}°")
````

````python
# Cross product (3D only)
cross = np.cross(a, b)
print("Cross product a×b:", cross)
print("Dot with a (should be 0):", np.dot(cross, a))  # perpendicular check
print("Dot with b (should be 0):", np.dot(cross, b))
````

## 2. Matrices

````python
# Creating matrices
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("A =\n", A)
print("shape:", A.shape)
````

````python
# Special matrices
print("Identity (3x3):\n", np.eye(3))
print("\nZeros (2x3):\n", np.zeros((2, 3)))
print("\nDiagonal:\n", np.diag([3, 7, 4]))
````

````python
# Matrix operations
print("A + B:\n", A + B)
print("\nA - B:\n", A - B)
print("\n3 * A:\n", 3 * A)
print("\nElement-wise A * B:\n", A * B)
print("\nMatrix multiply A @ B:\n", A @ B)
````

                                                                                                                                                        
```
● Element-wise multiplication (*) is used when you want to scale each element independently — not combine rows and columns.

Common uses:

1. Applying a mask / weights
data   = np.array([[10, 20], [30, 40]])
mask   = np.array([[1, 0], [0, 1]])   # keep diagonal only

data * mask  →  [[10, 0], [0, 40]]

2. Neural networks — applying activation element-wise
weights = np.array([0.5, 0.2, 0.8])
inputs  = np.array([1.0, 2.0, 3.0])

weights * inputs   # scale each input by its weight independently
→ [0.5, 0.4, 2.4]  

3. Image processing
image  = np.array([[255, 200], [100, 50]])
filter = np.array([[0.5, 0.5], [0.5, 0.5]])  # darken

image * filter  →  [[127.5, 100], [50, 25]]

---
Rule of thumb:

┌─────────────────────────────┬──────────────────────────────────┐
│     Use @ (matrix mul)      │       Use * (element-wise)       │
├─────────────────────────────┼──────────────────────────────────┤
│ Transforming vectors/space  │ Scaling individual values        │
├─────────────────────────────┼──────────────────────────────────┤
│ Linear algebra operations   │ Applying weights, masks, filters │
├─────────────────────────────┼──────────────────────────────────┤
│ Combining row × column info │ Each element is independent      │
└─────────────────────────────┴──────────────────────────────────┘
```

````python
# Transpose
print("A =\n", A)
print("\nA transpose:\n", A.T)
````

````python
# Determinant
print("det(A) =", np.linalg.det(A))

singular = np.array([[1, 2], 
                     [2, 4]])
print("det(singular) =", np.linalg.det(singular))   # should be 0
````

````python
# Inverse
A_inv = np.linalg.inv(A)
print("A inverse:\n", A_inv)
print("\nA @ A_inv (should be I):\n", np.round(A @ A_inv))
````

````python
# Solve system of linear equations: Ax = b
# 2x + y = 8
# 5x + 3y = 21
A_sys = np.array([[2, 1], [5, 3]])
b_sys = np.array([8, 21])

x = np.linalg.solve(A_sys, b_sys)
print("Solution x =", x)
print("Verify A @ x =", A_sys @ x)   # should equal b
````
