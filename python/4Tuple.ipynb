# Tuples

```python
t = (1, 2, 3, 2, 4)

# --- Properties ---
# ordered, immutable (cannot change after creation), allows duplicates

# --- Indexing and Slicing (same as list) ---
t[0]          # 1
t[-1]         # 4
t[1:3]        # (2, 3)
t[::-1]       # (4, 2, 3, 2, 1)  reverse

# --- Methods (only 2 — immutable so no add/remove) ---
t.count(2)    # 2  — count occurrences
t.index(3)    # 2  — index of first occurrence

# --- Unpacking ---
a, b, c, d, e = t
print(a, b, c)

# partial unpacking
first, *rest = t
print(first, rest)    # 1  [2, 3, 2, 4]

# --- Single element tuple — must have trailing comma ---
single = (5,)         # tuple
not_tuple = (5)       # just int!
print(type(single), type(not_tuple))
```
