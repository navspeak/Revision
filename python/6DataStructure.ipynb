# Data Structures — Stack, Queue

```python
from collections import deque

# --- Stack (LIFO) — use list ---
stack = []
stack.append(1)   # push
stack.append(2)
stack.append(3)
stack.pop()       # 3 — pop from top
print("Stack top:", stack[-1])

# --- Queue (FIFO) — use deque ---
queue = deque()
queue.append(1)    # enqueue
queue.append(2)
queue.append(3)
queue.popleft()    # 1 — dequeue from front
print("Queue front:", queue[0])

# --- Why deque for queue, not list? ---
# list.pop(0) is O(n) — shifts all elements
# deque.popleft() is O(1) — optimised for front removal
```
