```python
s = "Hello, World!"

# --- Indexing and Slicing ---
s[0]          # 'H'       first char
s[-1]         # '!'       last char
s[0:5]        # 'Hello'   start:stop (stop excluded)
s[7:]         # 'World!'  from index 7 to end
s[:5]         # 'Hello'   from start to 5 (excluded)
s[::2]        # every 2nd char
s[::-1]       # 'reverse the string'
s[7:12]       # 'World'

print(s[0], s[-1], s[0:5], s[::-1])
```

```python
from functools import reduce
s = "racecar"
three = []
four = []
n = len(s)
for i in range(n):
   if (i+5) <= n:
     four.append(s[i:i+5]) 
   if (i+4) <= n:
     three.append(s[i:i+4]) 

print(three)
print(four)

all = three + four
print(all)

palindromes = list(filter(lambda w: w == w[::-1], all))
print(palindromes)
points = {4: 5, 5: 10}
x = reduce(lambda acc, w: acc + points.get(len(w), 0), palindromes, 0)
print(x)
```

```python
# --- Generator Expression ---
# like list comprehension but LAZY — produces one value at a time, no list built in memory
# syntax: (expression for item in iterable if condition)  — round brackets

nums = "a1b2c3d4"

# list comprehension — builds full list immediately
digits_list = [ch for ch in nums if ch.isdigit()]   # ['1', '2', '3', '4']
digits_list2 = list(filter(lambda ch: ch.isdigit(),nums))
digits_list3 = list(filter(str.isdigit,nums))
print(digits_list)
print(digits_list2)
print (digits_list == digits_list2)

# generator expression — lazy, produces items on demand
digits_gen = (ch for ch in nums if ch.isdigit())    # <generator object>

# ''.join() accepts any iterable — generator OR list — so no need to build a list
result = ''.join(ch for ch in nums if ch.isdigit())  # '1234'
result2 = ''.join(digits_gen)  # '1234'
result3 = ''.join(digits_list)  # '1234'
print(result)
print(result2)
print(result3)
# ch is a str (single character string); join concatenates them

# why use generator over list?
# — no intermediate list → less memory (matters for large strings)
# — join only needs to iterate once, so lazy is fine

# string comprehension doesn't exist — use ''.join() with a generator/list instead
print(digits_list, result)
```
