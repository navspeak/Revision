```python
s = {1, 2, 3, 4}

# --- Properties ---
# unordered, no duplicates, no indexing

# --- Methods ---
s.add(5)               # add one item
s.update([6, 7])       # add multiple items
s.remove(1)            # remove — KeyError if not found
s.discard(99)          # remove — no error if not found
s.pop()                # remove and return arbitrary item
s.copy()               # shallow copy
s.clear()              # empty the set

# --- Set operations ---
a = {1, 2, 3}
b = {3, 4, 5}

a | b                  # {1,2,3,4,5}   union
a & b                  # {3}           intersection
a - b                  # {1,2}         difference (in a, not in b)
a ^ b                  # {1,2,4,5}     symmetric difference (not in both)

a.issubset(b)          # False
a.issuperset(b)        # False
a.isdisjoint(b)        # False  (they share 3)

print(a | b, a & b, a - b)
```

```python
# --- Set Comprehension ---
# {expression for item in iterable if condition}

squares = {x**2 for x in range(6)}          # {0, 1, 4, 9, 16, 25} — order not guaranteed
evens   = {x for x in range(10) if x % 2 == 0}  # {0, 2, 4, 6, 8}

# useful: deduplicate a list while filtering
words = ["apple", "banana", "apple", "cherry", "banana"]
unique_long = {w for w in words if len(w) > 5}   # {'banana', 'cherry'}

print(squares, evens, unique_long)
```
