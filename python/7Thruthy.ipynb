- Falsy values (evaluate to False):
    - False
    - None
    - 0, 0.0, 0j           # any numeric zero
    - "", [], (), {}, set()  # empty sequences/collections
    - range(0)
- Everything else is truthy, including:
    - "False"   # non-empty string, even if it says "False"
    - "0"       # non-empty string
    - [0]       # list with one element (even if that element is falsy)
    - {0: 0}    # non-empty dict
    - -1, 0.1   # any non-zero number
