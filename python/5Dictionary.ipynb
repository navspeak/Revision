# Dictionaries

```python
d = {'name': 'Alice', 'age': 30, 'city': 'Delhi'}

# --- Access ---
d['name']              # 'Alice'  — KeyError if missing
d.get('name')          # 'Alice'  — None if missing (safe)
d.get('phone', 'N/A')  # 'N/A'   — default if missing

# --- Check key ---
'name' in d            # True
'phone' in d           # False
'phone' not in d       # True

# --- Methods ---
d.keys()               # dict_keys(['name', 'age', 'city'])
d.values()             # dict_values(['Alice', 30, 'Delhi'])
d.items()              # dict_items([('name','Alice'), ...])

d.update({'age': 31})  # update existing key
d.update({'email': 'a@b.com'})  # add new key
d.pop('city')          # remove and return value
d.setdefault('score', 0)  # set key only if not already present
d.copy()               # shallow copy
d.clear()              # empty

# --- Iteration ---
for k, v in d.items():
    print(k, v)
```

```python
# --- Dict Comprehension ---
# {key_expr: val_expr for item in iterable if condition}

squares = {x: x**2 for x in range(5)}       # {0:0, 1:1, 2:4, 3:9, 4:16}
inverted = {v: k for k, v in {"a": 1, "b": 2}.items()}  # {1:'a', 2:'b'}

# filter dict — keep only high scorers
scores = {"Alice": 85, "Bob": 42, "Carol": 91}
passed = {k: v for k, v in scores.items() if v >= 60}   # {'Alice':85,'Carol':91}

print(squares, passed)
```

```python
from collections import Counter, defaultdict

# --- Counter ---
# counts occurrences of items in an iterable
words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
c = Counter(words)
# Counter({'apple': 3, 'banana': 2, 'cherry': 1})

c["apple"]          # 3
c["missing"]        # 0  — no KeyError (unlike regular dict)
c.most_common(2)    # [('apple', 3), ('banana', 2)]
c.most_common()     # all items sorted by count descending

# works on strings too
Counter("banana")   # Counter({'a': 3, 'n': 2, 'b': 1})

# --- defaultdict ---
# like a dict but returns a default value for missing keys instead of raising KeyError
dd = defaultdict(list)     # default factory: list → missing key returns []
dd["fruits"].append("apple")
dd["fruits"].append("banana")
dd["vegs"].append("carrot")
# defaultdict(list, {'fruits': ['apple', 'banana'], 'vegs': ['carrot']})

dd2 = defaultdict(int)     # default factory: int → missing key returns 0
for word in ["a", "b", "a", "c", "a", "b"]:
    dd2[word] += 1
# defaultdict(int, {'a': 3, 'b': 2, 'c': 1})  — same as Counter but manual

print(c, dict(dd), dict(dd2))
```
