# Lists

```python
lst = [1, 2, 3, 4, 5]

# --- Slicing ---
lst[0]        # 1
lst[-1]       # 5
lst[1:3]      # [2, 3]
lst[::-1]     # [5, 4, 3, 2, 1]  reverse

# --- Methods ---
lst.append(6)          # [1,2,3,4,5,6]        add one item
lst.extend([7, 8])     # [1,2,3,4,5,6,7,8]    add multiple items
lst.insert(0, 0)       # insert 0 at index 0
lst.pop()              # removes and returns last item
lst.pop(0)             # removes and returns item at index 0
lst.remove(3)          # removes first occurrence of value 3
lst.index(4)           # returns index of value 4
lst.count(2)           # count occurrences of 2
lst.sort()             # sort in place
lst.reverse()          # reverse in place
lst.copy()             # shallow copy
lst.clear()            # empty the list

print(lst)
```

```python
# --- List Comprehension ---
# [expression for item in iterable if condition]

squares = [x**2 for x in range(5)]          # [0, 1, 4, 9, 16]
evens   = [x for x in range(10) if x % 2 == 0]  # [0, 2, 4, 6, 8]
upper   = [c.upper() for c in "hello"]      # ['H', 'E', 'L', 'L', 'O']

# nested comprehension — flatten 2D list
matrix = [[1, 2], [3, 4], [5, 6]]
flat = [x for row in matrix for x in row]   # [1, 2, 3, 4, 5, 6]

print(squares, evens, flat)
```

```python
a = [1, 2, 3]

b1 = a.copy() # shallow copy
b2 = a[:]
b3 = list(a)

b1 == b2 == b3 == a    # True  (same contents)
b1 is a                # False (different object)
b2 is a                # False
b3 is a                # False
```

```python
a = [[1, 2], [3, 4]]
b1 = a.copy()
b2 = a[:]
b3 = list(a)

b1[0] is a[0]   # True — inner list still shared
b2[0] is a[0]   # True
b3[0] is a[0]   # True
```

```python
# map(func, iterable) applies func to every element of iterable. 
# It returns a map object (a lazy iterator), not a list — so you wrap it in list() to actually materialize the results.
list(map(len, ["hi", "hello", "hey"]))   # [2, 5, 3]
list(map(abs, [-1, -2, 3]))              # [1, 2, 3]
# These two are equivalent:
[str(x) for x in a]
list(map(str, a))

# These two are equivalent:
[x * 2 for x in a]
list(map(lambda x: x * 2, a))
```
